```python
"""
Quick EDA: Understand what we downloaded
"""

import pandas as pd
import os
import glob

# Find all CSV files
data_dir = '/workspace/data/raw'
all_csvs = glob.glob(f'{data_dir}/**/*.csv', recursive=True)

print(f"Found {len(all_csvs)} CSV files\n")

# Quick peek at each
for csv_path in all_csvs:
    print("=" * 60)
    print(f"FILE: {os.path.basename(csv_path)}")
    print(f"PATH: {csv_path}")
    print("=" * 60)
    
    try:
        df = pd.read_csv(csv_path, nrows=5)
        print(f"Shape: {df.shape[0]} rows (sample), {df.shape[1]} columns")
        print(f"\nColumns: {', '.join(df.columns.tolist())}")
        print(f"\nFirst 3 rows:")
        print(df.head(3))
        print("\n")
    except Exception as e:
        print(f"Error reading: {e}\n")
```
